```python
# test
nlp = spacy.load('de_core_news_sm') # load the german model
doc = nlp("Die Eichhörnchen (Sciurus) sind eine Gattung der Baumhörnchen (Sciurini) innerhalb der Familie der Hörnchen (Sciuridae). Ein auffälliges Merkmal ist der hochgestellte buschige Schwanz. Die in Mitteleuropa bekannteste Art ist das Eurasische Eichhörnchen, das gemeinhin einfach als Eichhörnchen bezeichnet wird. Alle Eichhörnchen sind Waldbewohner und ernähren sich primär von Samen und Früchten. Die meisten Arten sind auf dem amerikanischen Doppelkontinent beheimatet. Nur vier der 30 Arten leben in der Alten Welt, sie sind über Europa, Vorder-, Nord- und Ostasien verbreitet. Eine weitere Art (Grauhörnchen) ist als Neozoon in Teilen Europas eingebürgert.")
# above from wikipedia, eichhornchen
for token in doc:
    print(token.text, token.lemma_, token.pos_, token.tag_, token.dep_,
            token.shape_, token.is_alpha, token.is_stop)
```
